```python
df_alumnos_ds_total = df_alumnos = pd.concat([df_alumnos_ds_01, df_alumnos_ds_02], ignore_index=True)
df_alumnos_ds_total
```

```python
# Convertir fecha a formato YYYY-MM-DD
df_alumnos_ds_total['Fecha_comienzo'] = pd.to_datetime(
    df_alumnos_ds_total['Fecha_comienzo'], 
    dayfirst=True
).dt.date

# Columnas de proyectos (solo para identificar bootcamp por proyecto)
proyectos_cols = ['Proyecto_HLF','Proyecto_EDA','Proyecto_BBDD','Proyecto_ML','Proyecto_Deployment']

# Pasar de ancho a largo
df_long = df_alumnos_ds_total.melt(
    id_vars=['Nombre','Email','Promoción','Fecha_comienzo','Campus'],
    value_vars=proyectos_cols,
    var_name='Proyecto',
    value_name='Calificacion'
)

# Bootcamp según proyecto (DS = 0)
bootcamp_map = {
    'Proyecto_HLF': 0,
    'Proyecto_EDA': 0,
    'Proyecto_BBDD': 0,
    'Proyecto_ML': 0,
    'Proyecto_Deployment': 0
}

# Promoción → ID
promocion_map = {
    'Febrero': 0,
    'Marzo': 1,
    'Abril': 2,
    'Junio': 3,
    'Septiembre': 4,
    'Octubre': 5
}

# Campus → ID
campus_map = {
    'Madrid': 0,
    'Valencia': 1,
    'Bilbao': 2,
    'Vitoria': 3,
    'Sevilla': 4
}

# Aplicar mapas
df_long['id_bootcamp'] = df_long['Proyecto'].map(bootcamp_map)
df_long['id_promocion'] = df_long['Promoción'].map(promocion_map)
df_long['id_campus'] = df_long['Campus'].map(campus_map)

# Crear id_alumno único empezando por 0
df_long['id_alumno'] = range(len(df_long))

# Seleccionar solo las columnas que existen en la tabla Alumno
df_final_alumnos_ds = df_long[
    ['id_alumno', 'Nombre', 'Email', 'id_bootcamp', 'id_promocion', 'id_campus', 'Fecha_comienzo']
]

# Renombrar columnas
df_final_alumnos_ds = df_final_alumnos_ds.rename(columns={
    'Nombre': 'nombre',
    'Email': 'email',
    'Fecha_comienzo': 'fecha_inicio'
})

df_final_alumnos_ds
```

```python
df_final_alumnos_ds.to_sql("alumno", engine, if_exists="append", index= False)
```

```python
df_proy_ds = df_long[['Calificacion']].copy()  
df_proy_ds['id_alumno'] = df_final_alumnos_ds['id_alumno'].values
df_proy_ds['id_bootcamp'] = 0  # Bootcamp DS

df_proy_ds['Calificacion'] = pd.to_numeric(df_proy_ds['Calificacion'], errors='coerce')
df_proy_ds['calificacion'] = df_proy_ds['Calificacion'].apply(
    lambda x: 'apto' if x >= 5 else 'no apto'
)

df_proy_fs = df_long_fs[['Calificacion']].copy()  # df_long_fs es FS
df_proy_fs['id_alumno'] = df_final_alumnos_fs['id_alumno'].values
df_proy_fs['id_bootcamp'] = 1  # Bootcamp FS

df_proy_fs['Calificacion'] = pd.to_numeric(df_proy_fs['Calificacion'], errors='coerce')
df_proy_fs['calificacion'] = df_proy_fs['Calificacion'].apply(
    lambda x: 'apto' if x >= 5 else 'no apto'
)

df_proyectos_realizados = pd.concat([df_proy_ds, df_proy_fs], ignore_index=True)

df_proyectos_realizados['id_proyectos_realizados'] = range(len(df_proyectos_realizados))

df_proyectos_realizados = df_proyectos_realizados[
    ['id_proyectos_realizados', 'id_alumno', 'id_bootcamp', 'calificacion']
]
```
